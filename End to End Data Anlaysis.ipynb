```python
#read data from the file and handle the null values
import pandas as pd
df = pd.read_csv('orders.csv',na_values = ['Not Available','unknown'])
df['Ship Mode'].unique()
```

```python
#Rename columns names and make them lower case and replace space with underscore
df.rename(columns={'Order Id' : 'order id'})
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df.head(5)
```

```python
#Drive new columns discount, sale price and profit
df['discount']=df['list_price']*df['discount_percent']*.01
df['sale_price']= df['list_price']-df['discount']
df['profit']=df['sale_price']-df['cost_price']
df.head(5)
```

```python
#convert order date from object data type to datetime
df.dtypes
df['order_date']=pd.to_datetime(df['order_date'],format="%Y-%m-%d")
df.dtypes
```

```python
import pandas as pd
from dotenv import load_dotenv
import os
import mysql.connector

# Load environment variables from .env file
load_dotenv()

# Database credentials
host = 'localhost'
database = 'pythonsql'
user = os.getenv('DB_USER')
password = os.getenv('DB_PASSWORD')

# Establish connection
connection = mysql.connector.connect(
    host=host,
    database=database,
    user=user,
    password=password
)

# Check if the connection is successful
if connection.is_connected():
    print("Connected to MySQL server")

    # Create a cursor object
    cursor = connection.cursor()

    # Assuming df is your already existing DataFrame
    # Example: df = pd.DataFrame(...)

    # Handle NaN values: Replace NaN with None (which represents NULL in MySQL)
    df = df.where(pd.notnull(df), None)

    # Define column names and their respective SQL data types
    column_type_mapping = {
        "order_id": "INT PRIMARY KEY",
        "order_date": "DATE",
        "ship_mode": "VARCHAR(20)",
        "segment": "VARCHAR(20)",
        "country": "VARCHAR(20)",
        "city": "VARCHAR(20)",
        "state": "VARCHAR(20)",
        "postal_code": "VARCHAR(20)",
        "region": "VARCHAR(20)",
        "category": "VARCHAR(20)",
        "sub_category": "VARCHAR(20)",
        "product_id": "VARCHAR(50)",
        "quantity": "INT",
        "discount": "DECIMAL(7,2)",
        "sale_price": "DECIMAL(7,2)",
        "profit": "DECIMAL(7,2)"
    }

    # Extract DataFrame column names
    columns = df.columns.tolist()

    # Assign correct column types, defaulting to VARCHAR(255) if not found in mapping
    column_definitions = [f"{col} {column_type_mapping.get(col, 'VARCHAR(255)')}" for col in columns]

    # Dynamically generate the CREATE TABLE query
    create_table_query = f"""
    CREATE TABLE IF NOT EXISTS df_orders (
        {', '.join(column_definitions)}
    );
    """

    # Create the table if it doesn't exist
    cursor.execute(create_table_query)

    # Generate the INSERT query dynamically
    insert_query = f"INSERT INTO df_orders ({', '.join(columns)}) VALUES ({', '.join(['%s'] * len(columns))})"

    # Loop through DataFrame and insert each row into the table
    for index, row in df.iterrows():
        cursor.execute(insert_query, tuple(row))

    # Commit the transaction
    connection.commit()

    print("Data inserted successfully!")

# Close the connection
connection.close()
```
